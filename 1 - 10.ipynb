## Add two numbers
##Take two numbers as input and display their sum, difference, product, and quotient.

```python
Enter_First_Number = int(input("Enter The First Number:- "))
Enter_Second_Number = int(input("Enter The Second Number:- "))
Choose_The_Operator = input("Enter The Operator, (+, -, *, /, //):- ")

if Choose_The_Operator == "+":
    print(f"The Sum of Two Number is:- {Enter_First_Number + Enter_Second_Number}")
elif Choose_The_Operator == "-":
    print(f"The Subtraction of The Number is:- {Enter_First_Number - Enter_Second_Number}")
elif Choose_The_Operator == "*":
    print(f"The Multiplication of Two Number is:- {Enter_First_Number * Enter_Second_Number}")
elif Choose_The_Operator == "/":
    print(f"The Division of Two Number is:- {Enter_First_Number / Enter_Second_Number}")
elif Choose_The_Operator == "//":
    print(f"The Floor Division of Two Number is:- {Enter_First_Number // Enter_Second_Number}")
else:
    print("Sorry............\n I Think You Choose a Wrong Operator OR a Number(Instead of Number Like Float,Bool OR etc.....) \n So Please Try Again.....")
```

## Largest of three numbers
Find the largest of three numbers without using max().

```python
First_Number = int(input("Enter The First Number:- "))
Second_Number = int(input("Enter The Second Number:- "))
Third_Number = int(input("Enter The Third Number:- "))

if First_Number > Second_Number:
    print(f"First Number {First_Number} is Largest Above The All Three Numbers")
elif Second_Number > Third_Number:
    print(f"Second Number {Second_Number} is Largest Above The All Three Numbers")
elif Third_Number > First_Number:
    print(f"Third Number {Third_Number} is Largest Above The All Three Numbers")
else:
    print("Try Again......... \n I Think You Enter The Equal Numbers")
```

## Leap year checker
Check whether a year is a leap year.

```python
year = int(input("Enter a year: "))

# A leap year is divisible by 4, but century years must also be divisible by 400
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a Leap Year!")
else:
    print(f"{year} is a Regular Year.")
```

## Multiplication table
Print the multiplication table of a number from 1 to 10.

```python
Enter_The_Number = int(input("Enter The Number,Which You Want The Table"))

for i in range(1,11):
    result = Enter_The_Number * i

    print(f"{Enter_The_Number} * {i} = {result}")
```

## Sum of natural numbers
Calculate the sum of the first n natural numbers using a loop.

```python
Enter_The_Natural_Number = int(input("Enter The Natural Number:- "))

Total_Sum = 0

for i in range(1,Enter_The_Natural_Number + 1):
    Total_Sum = Total_Sum + i
    

print(f"The Sum of {Enter_The_Natural_Number} Natural Number is {Total_Sum}.")
```

## Reverse a number
Reverse an integer without converting it to a string.

```python
Num = int(input("Enter The Number, which You Wants To Reverse"))

N = 0

while Num > 0:
    Last_Digit = Num % 10
    N = (N * 10) + Last_Digit
    Num = Num // 10

print(f"The Number {Num} Is Reveresd as {N}")
```

## Prime number checker
Determine whether a given integer is prime.

```python
Num = int(input("Enter The Number, Which is Prime or Not Prime:- "))

if Num <= 1:
    print(f"{Num} is 1 or Less Than One.\n So Enter The Number Which Is Greater Than One.")
else:
    for i in range(2,Num):
        if Num % i == 0:
            print(f"{Num} is Not a Prime Number, It's Divisible By {i}")

            break

    else:
        print(f"{Num} is Prime")
```
